# Menghapus Missing Value Untuk Semua Baris

```python
# MENGHAPUS MISSING VALUE

df = df.dropna(how="all").copy()

print("Jumlah baris setelah menghapus baris kosong:", len(df))
```

# Treatment Kolom Teks

```python
text_columns = [
    "ssr",
    "bulan",
    "kab_kota",
    "justifikasi",
    "kategori_bl"
]

# Hanya proses kolom teks yang benar-benar ada di data
active_text_columns = [col for col in text_columns if col in df.columns]

for col in active_text_columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )
```

```python
if active_text_columns:
    df[active_text_columns] = df[active_text_columns].replace(
        {
            "": pd.NA,
            "-": pd.NA,
            "–": pd.NA,
            "—": pd.NA,
            "nan": pd.NA,
            "None": pd.NA
        }
    )
```

# Treatment Kolom Numeric

```python
numeric_columns = [
    "bl",
    "jml_kegiatan_planning",
    "unit_cost",
    "total_budget",
    "jml_kegiatan_realisasi",
    "realisasi",
    "persentase_budget"
]

# Hanya proses kolom numerik yang ada di data
active_numeric_columns = [col for col in numeric_columns if col in df.columns]
```

```python
# MEMPERBAIKI PENULISAN ANGKA PADA KOLOM NUMERIK

def clean_number(value):

    if pd.isna(value):
        return np.nan

    if isinstance(value, (int, float, np.integer, np.floating)):
        return float(value)

    value = str(value).strip()

    if value in ["", "-", "–", "—", "nan", "None"]:
        return np.nan

    value = value.replace("%", "")
    value = value.replace(" ", "")

    # Contoh:
    # 1.234.567,89 -> 1234567.89
    if "," in value and "." in value:

        if value.rfind(",") > value.rfind("."):
            value = value.replace(".", "")
            value = value.replace(",", ".")
        else:
            value = value.replace(",", "")

    # Contoh:
    # 1,234,567
    # atau
    # 123,45
    elif "," in value:

        parts = value.split(",")

        if len(parts[-1]) <= 2:
            value = value.replace(",", ".")
        else:
            value = value.replace(",", "")

    # Contoh:
    # 1.234.567
    elif value.count(".") > 1:
        value = value.replace(".", "")

    try:
        return float(value)

    except:
        return np.nan
```

```python
for col in active_numeric_columns:
    df[col] = df[col].apply(clean_number)
```

# Memperbaiki Kolom "Bulan"

```python
month_map = {
    "jan": 1,
    "januari": 1,

    "feb": 2,
    "februari": 2,

    "mar": 3,
    "maret": 3,

    "apr": 4,
    "april": 4,

    "mei": 5,
    "may": 5,

    "jun": 6,
    "juni": 6,

    "jul": 7,
    "juli": 7,

    "agu": 8,
    "ags": 8,
    "aug": 8,
    "agustus": 8,

    "sep": 9,
    "sept": 9,
    "september": 9,

    "okt": 10,
    "oct": 10,
    "oktober": 10,

    "nov": 11,
    "november": 11,

    "des": 12,
    "dec": 12,
    "desember": 12
}
```

```python
def get_month_number(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    # Hilangkan titik
    value = value.replace(".", " ")

    # Rapikan spasi
    value = " ".join(value.split())

    for key, number in month_map.items():

        if key in value:
            return number

    return np.nan
```

```python
df["bulan_no"] = df["bulan"].apply(get_month_number)
```

```python
df[["bulan", "bulan_no"]].drop_duplicates().sort_values("bulan_no")
```

```python
month_names = {
    1: "Januari",
    2: "Februari",
    3: "Maret",
    4: "April",
    5: "Mei",
    6: "Juni",
    7: "Juli",
    8: "Agustus",
    9: "September",
    10: "Oktober",
    11: "November",
    12: "Desember"
}

df["bulan"] = df["bulan_no"].map(month_names)
```

```python
df["tahun"] = 2026
```

```python
df["tanggal_bulan"] = pd.to_datetime(
    {
        "year": df["tahun"],
        "month": df["bulan_no"],
        "day": 1
    },
    errors="coerce"
)
```

```python
df["periode"] = df["tanggal_bulan"].dt.strftime("%Y-%m")
```

# Isi Nilai Numerik dengan Angka 0

```python
zero_columns = [
    "bl",
    "jml_kegiatan_planning",
    "unit_cost",
    "total_budget",
    "jml_kegiatan_realisasi",
    "realisasi"
]

for col in zero_columns:
    if col in df.columns:
        df[col] = df[col].fillna(0)
```

# Validasi TOTAL BUDGET

```python
df["total_budget_valid"] = (
    df["jml_kegiatan_planning"] *
    df["unit_cost"]
)
```

```python
df["selisih_budget"] = (
    df["total_budget"] -
    df["total_budget_valid"]
)
```

```python
df[
    df["selisih_budget"].abs() > 1
][
    [
        "bl",
        "jml_kegiatan_planning",
        "unit_cost",
        "total_budget",
        "total_budget_valid",
        "selisih_budget"
    ]
].head(20)
```

# Hitung Ulang Persentase Serapan

```python
df["persentase_valid"] = np.where(
    df["total_budget_valid"] > 0,
    df["realisasi"] / df["total_budget_valid"],
    0
)
```

# Buat Flag Untuk Data yang Perlu Dicek

```python
df["flag_realisasi"] = np.select(

    [
        df["realisasi"] < 0,

        df["realisasi"] > df["total_budget_valid"],

        df["realisasi"] == df["total_budget_valid"]
    ],

    [
        "Realisasi negatif",

        "Realisasi > Budget",

        "Realisasi = Budget"
    ],

    default="Normal"
)
```
